````python
with open("./descrition.csv") as f:
    content = f.readlines()
print(content[:10])
````

````python
# Describe data
````

````python
print("columns: ", len(content[0].split("|")))
print("lines: ",len(content))
words = 0
for line in content:
    words += len(line.split(" "))
print("words: ", words)
````

````python
# get data from positions
positions = []
for line in content[1:]:
    positions.append(line.split(",")[0])
print("positions", positions[:10]) 
````

````python
# count words
occurrences = {}
for line in positions:
    words = line.split(" ")
    for word in words:
        occurrences[word] = occurrences.get(word, 0) + 1
#print("word count", word_counts)
for word, count in sorted(occurrences.items(), key=lambda item: item[1], reverse=True):
    print(f"'{word}': {count}")
````
